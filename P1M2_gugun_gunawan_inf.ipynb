### 10. Model Inference

##### 1. Load Libraries & Model

```python
import pandas as pd
import joblib

# Load model pipeline yang sudah disimpan (pastikan file .pkl ada di folder yang sama)
model = joblib.load('best_rf_credit_model.pkl')
```

##### 2. Creating New Data (Data Dummy)
Kita buat 2 contoh nasabah:

Nasabah A: Gaji besar, pinjaman kecil, rumah milik sendiri (Prediksi harusnya Aman).

Nasabah B: Gaji kecil, pinjaman besar, bunga tinggi, ada histori telat bayar (Prediksi harusnya Berisiko).

```python
# Data dalam format asli (sebelum preprocessing)
data_baru = {
    'person_age': [35, 22],
    'person_income': [120000, 15000],
    'person_home_ownership': ['MORTGAGE', 'RENT'],
    'person_emp_length': [10.0, 1.0],
    'loan_intent': ['VENTURE', 'PERSONAL'],
    'loan_grade': ['A', 'E'],
    'loan_amnt': [5000, 25000],
    'loan_int_rate': [7.5, 18.5],
    'loan_percent_income': [0.04, 0.60],
    'cb_person_default_on_file': ['N', 'Y'],
    'cb_person_cred_hist_length': [12, 2]
}

df_inf = pd.DataFrame(data_baru)
```

### 3. Prediction

```python
# Melakukan prediksi menggunakan pipeline
# Pipeline akan otomatis melakukan Scaling dan Encoding
y_pred_inf = model.predict(df_inf)
y_pred_proba = model.predict_proba(df_inf)[:, 1] # Probabilitas gagal bayar

# Memasukkan hasil ke DataFrame agar mudah dibaca
df_inf['Prediction'] = y_pred_inf
df_inf['Status'] = df_inf['Prediction'].map({0: 'Approved (Low Risk)', 1: 'Rejected (High Risk)'})
df_inf['Risk_Probability'] = y_pred_proba

# Menampilkan hasil
df_inf[['person_income', 'loan_amnt', 'loan_grade', 'Status', 'Risk_Probability']]
```

### Insight
Model berhasil memprediksi dua data baru dengan logis; nasabah dengan profil finansial kuat diterima, sedangkan nasabah dengan rasio hutang tinggi dan histori default ditolak.
